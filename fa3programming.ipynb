```python
# Input
name = input("Enter student name: ")
score = float(input("Enter score: "))

# Output
if 90 <= score <= 100:
    grade = "A"
elif 80 <= score <= 89:
    grade = "B"
elif 70 <= score <= 79:
    grade = "C"
else:
    grade = "F"

print("Name:", name)
print("Grade:", grade)
```
